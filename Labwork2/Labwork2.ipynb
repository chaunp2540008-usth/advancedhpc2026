Run the following cell to verify that numba can use Cuda...

```python
%%python
from numba import cuda

def bool_to_str(predicate: bool) -> str:
    if predicate:
        return 'yes'
    return 'no'

def is_cuda_available_str() -> str:
    return bool_to_str(cuda.is_available())

print(f'is cuda available? {is_cuda_available_str()}')
```

Run the following cell for a short test of numba/cuda...

```python
%%python
from numba import cuda
device = cuda.get_current_device()

gpu_name = device.name.decode('utf-8')
mp_count = device.MULTIPROCESSOR_COUNT
cc_major, cc_minor = device.compute_capability

print(f'GPU name: {gpu_name}')
print(f'Multiprocessor count: {mp_count}')
print(f'Compute capability: {cc_major}.{cc_minor}')
print(f'Core count: {mp_count * 64}')  # 64 cores per multiprocessor for 7.5


ctx = cuda.current_context()
free_mem, total_mem = ctx.get_memory_info()

print(f'Free memory: {free_mem / 1024 ** 2:.2f} MB')
print(f'Total memory: {total_mem / 1024 ** 2:.2f} MB')
```
